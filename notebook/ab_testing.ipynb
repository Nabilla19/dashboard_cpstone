# A/B Testing
# Perbandingan Dataset Clean vs Feature Engineering

Tujuan:
Menguji apakah feature engineering dapat meningkatkan performa model machine learning.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Membuat dataframe hasil eksperimen
ab_results = pd.DataFrame({
    'Model': ['A - Clean Data', 'B - Feature Engineering'],
    'Accuracy': [86.27, 86.47],
    'MAE': [0.6875, 0.6550]
})

print(ab_results)

# Visualisasi Accuracy
plt.figure(figsize=(6,4))
plt.bar(ab_results['Model'], ab_results['Accuracy'])
plt.title('Perbandingan Accuracy')
plt.ylabel('Accuracy (%)')
plt.show()

# Visualisasi MAE
plt.figure(figsize=(6,4))
plt.bar(ab_results['Model'], ab_results['MAE'])
plt.title('Perbandingan MAE')
plt.ylabel('MAE')
plt.show()
```

# Insight A/B Testing

## Insight 1 — Feature Engineering meningkatkan performa model
Hasil pengujian menunjukkan bahwa dataset hasil feature engineering menghasilkan accuracy lebih tinggi dibandingkan dataset clean biasa.

- Clean Data Accuracy: 86.27%
- Feature Engineering Accuracy: 86.47%

Ini menunjukkan bahwa penambahan fitur baru membantu model mengenali pola data dengan lebih baik.

---

## Insight 2 — Nilai MAE lebih rendah pada Feature Engineering
MAE (Mean Absolute Error) pada model feature engineering lebih kecil dibandingkan model clean data.

- Clean Data MAE: 0.6875
- Feature Engineering MAE: 0.6550

Semakin kecil MAE, semakin dekat hasil prediksi model dengan nilai sebenarnya.

---

## Insight 3 — Peningkatan performa model
Feature engineering memberikan peningkatan accuracy sebesar 0.20% serta menurunkan nilai MAE. Hal ini menunjukkan bahwa fitur tambahan membantu model memahami pola data mental health secara lebih optimal.

## Kesimpulan
Berdasarkan hasil A/B Testing, model Feature Engineering dipilih sebagai model terbaik karena menghasilkan accuracy lebih tinggi dan MAE lebih rendah dibandingkan model Clean Data.
